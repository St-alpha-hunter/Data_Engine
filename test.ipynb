```python
import os
import sys
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
print(os.getcwd())
path = Path.cwd() / "data" / "volume_data_1y"
print(path)
```

```python
df = pd.read_parquet(path / "20260507_211049.parquet")
df = df.reset_index()

##计算20日动量
df["mom_1_month"] = (
    df.groupby("symbol")["adjClose"]
    .pct_change(20)
)

df["volume_1_month"] = (
    df.groupby("symbol")["volume"]
    .pct_change(20)
)

df = df[df["date"]=="2026-03-03"]
df = df[["symbol","date","volume_1_month","mom_1_month"]]
df.head()

df["volume_1_month"] = (df["volume_1_month"]-df["volume_1_month"].min())/(df["volume_1_month"].max() - df["volume_1_month"].min())
df["mom_1_month"] = (df["mom_1_month"]-df["mom_1_month"].min())/(df["mom_1_month"].max() - df["mom_1_month"].min())
df["mom_1_month"].corr(df["volume_1_month"])

EX = df["mom_1_month"].mean()
print(EX)

EY = df["volume_1_month"].mean()
print(EY)


EXY = (
    df["mom_1_month"] *
    df["volume_1_month"]
).mean()

print(EXY)
```
